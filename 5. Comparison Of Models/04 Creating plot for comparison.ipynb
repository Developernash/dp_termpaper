```python
fig, axes = plt.subplots(
    2, 3,
    figsize=(18, 10),
    sharey=True
)

for ax, (file, title) in zip(axes.ravel(), files):
    data = np.load(file)

    old_vals = data["old_vals"]
    new_vals = data["new_vals"]

    ax.hist(
        old_vals,
        bins=20,
        alpha=0.5,
        density=True,
        label="Old model",
    )

    ax.hist(
        new_vals,
        bins=20,
        alpha=0.5,
        density=True,
        label="New model",
    )

    ax.axvline(
        old_vals.mean(),
        linestyle="--",
        label=f"Old mean: {old_vals.mean():.2f}"
    )

    ax.axvline(
        new_vals.mean(),
        linestyle="--",
        label=f"New mean: {new_vals.mean():.2f}"
    )

    ax.set_title(title)
    ax.set_xlabel("Criterion value")
    ax.legend()

axes[0, 0].set_ylabel("Density")
axes[1, 0].set_ylabel("Density")

fig.suptitle("Distribution of Criterion Values Across Seeds", fontsize=14)

plt.tight_layout()
plt.show()
```
